```python
# Ejercicio 3
from itertools import combinations

def factorial(n):
    """Calcula el factorial de un número."""
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
def combinacion(n, k):
    """Calcula el número combinatorio C(n, k)."""
    if n < k:
        return 0
    return factorial(n) // (factorial(k) * factorial(n - k))

# Solicitar al usuario la cantidad de variables
n = int(input("Ingrese la cantidad total de variables (n): "))

# Crear una lista de variables
variables = list(range(1, n + 1))

# Mostrar todas las combinaciones posibles de agrupaciones de a tres
todas_combinaciones_posibles = list(combinations(variables, 3))
print(f"Todas las variables son {variables}, las combinaciones de agrupaciones de a tres son: {todas_combinaciones_posibles}")

# Solicitar al usuario la cantidad de agrupaciones
k = 3

# Calcular el número combinatorio
resultado = combinacion(n, k)
print(f"El número combinatorio C( {n}, {k} ) o total de gráficos necesarios es: {resultado}")
```

```python
# Ejercicio 5
import matplotlib.pyplot as plt
import numpy as np

def dibujar_puntos(puntos):
    """
    Dibuja puntos en el plano 2D con etiquetas y leyenda.
    """
    colores = ['r', 'g', 'b']    
    for i, punto in enumerate(puntos):
        plt.scatter(*punto, color = colores[i], label = f'P{i+1}')
    plt.xlim(-1, max(puntos[0][0], puntos[1][0], puntos[2][0]) + 1)
    plt.ylim(-1, max(puntos[0][1], puntos[1][1], puntos[2][1]) + 1) 
    plt.xlabel('Eje X')
    plt.ylabel('Eje Y')
    plt.legend()
    plt.grid(True)
    plt.show()
    
# Ejemplo de uso con tres puntos:
punto1 = np.array([2, 3])
punto2 = np.array([4, 0])
punto3 = np.array([1, 5])
dibujar_puntos([punto1, punto2, punto3])
```

```python
# Ejercicio 7
import matplotlib.pyplot as plt
import numpy as np

# Función para dibujar puntos en el plano 2D
def dibujar_puntos(puntos):
    """
    Dibuja puntos en el plano 2D con etiquetas y leyenda.
    """
    colores = ['r', 'g', 'b']
    for i, punto in enumerate(puntos):
        plt.scatter(*punto, color=colores[i], label=f'P{i+1}')
    plt.xlabel('Eje X')
    plt.ylabel('Eje Y')
    plt.legend()
    plt.grid(True)
    plt.show()

#  Función para dibujar puntos en el espacio 3D
def dibujar_puntos3D(puntos):
    """
    Dibuja puntos en el espacio 3D con etiquetas y líneas desde el origen.
    """
    colores = ['r', 'g', 'b']
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    for i, punto in enumerate(puntos):
        ax.scatter(*punto, color=colores[i], label=f'P{i + 1}')

    # Dibujar líneas de los ejes desde el origen
    ax.plot([0, 0], [0, 0], [0, max(puntos[0][2], puntos[1][2], puntos[2][2]) + 1], color='black')  # Eje Z
    ax.plot([0, 0], [0, max(puntos[0][1], puntos[1][1], puntos[2][1]) + 1], [0, 0], color='black')  # Eje Y
    ax.plot([0, max(puntos[0][0], puntos[1][0], puntos[2][0]) + 1], [0, 0], [0, 0], color='black')  # Eje X
    ax.set_xlim([0, max(puntos[0][0], puntos[1][0], puntos[2][0]) + 1])
    ax.set_ylim([0, max(puntos[0][1], puntos[1][1], puntos[2][1]) + 1])
    ax.set_zlim([0, max(puntos[0][2], puntos[1][2], puntos[2][2]) + 1])
    ax.set_xlabel('Eje X')
    ax.set_ylabel('Eje Y')
    ax.set_zlabel('Eje Z')
    ax.legend()
    plt.show()

# Función para dibujar una recta en el plano 2D
def dibujar_recta(p1, p2):
    """
    Dibuja una recta en el plano 2D.
    """
    plt.plot([p1[0], p2[0]], [p1[1], p2[1]], color='black', linestyle='-', linewidth=2)
    plt.scatter(*p1, color='r', label='P1')
    plt.scatter(*p2, color='b', label='P2')
    plt.xlabel('Eje X')
    plt.ylabel('Eje Y')
    plt.legend()
    plt.grid(True)
    plt.show()

# Función para dibujar una recta en el espacio 3D
def dibujar_recta3D(p1, p2):
    """
    Dibuja una recta en el espacio 3D.
    """
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.plot([p1[0], p2[0]], [p1[1], p2[1]], [p1[2], p2[2]], color='black', linestyle='-', linewidth=2)
    ax.scatter(*p1, color='r', label='P1')
    ax.scatter(*p2, color='b', label='P2')
    ax.set_xlabel('Eje X')
    ax.set_ylabel('Eje Y')
    ax.set_zlabel('Eje Z')
    ax.legend()
    plt.show()

# Ejemplo de uso con puntos y una recta en 2D
punto1 = np.array([2, 3])
punto2 = np.array([4, 0])
punto3 = np.array([1, 5])
puntos = [punto1, punto2, punto3]
dibujar_puntos(puntos)
dibujar_recta(punto1, punto2)

# Ejemplo de uso con puntos y una recta en 3D
punto1_3d = np.array([2, 3, 1])
punto2_3d = np.array([4, 0, 2])
punto3_3d = np.array([1, 5, 3])
puntos_3d = [punto1_3d, punto2_3d, punto3_3d]
dibujar_puntos3D(puntos_3d)
dibujar_recta3D(punto1_3d, punto2_3d)
```

```python
# Ejercicio 8
import matplotlib.pyplot as plt
import numpy as np

def dibujar_curva2D():
    t = np.linspace(0, 2 * np.pi, 100)
    x = np.sin(t)
    y = np.cos(t)
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, label='Curva paramétrica: x = sin(t), y = cos(t)')
    plt.title('Curva en 2D')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)
    plt.axis('equal')  # Asegura que las escalas en x e y sean iguales
    plt.show()
def dibujar_curva3D():
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')
    t = np.linspace(0, 4 * np.pi, 100)
    x = np.sin(t)
    y = np.cos(t)
    z = t
    ax.plot(x, y, z, label='Curva paramétrica 3D: x = sin(t), y = cos(t), z = t')
    ax.set_title('Curva en 3D')
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('z')
    ax.legend()
    plt.show()
def dibujar_parabola():
    x = np.linspace(-10, 10, 400)
    y = x**2
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, label='Parábola: y = x^2')
    plt.title('Parábola en 2D')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)
    plt.axis('equal')
    plt.show()
    
# Dibujar las curvas
dibujar_curva2D()
dibujar_curva3D()
dibujar_parabola()
```

```python
# Ejercicio 9
import numpy as np
import matplotlib.pyplot as plt

# Función para graficar un plano dado por la ecuación general Ax + By + Cz + D = 0
def graficar_plano(A, B, C, D):
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    x = np.linspace(-10, 10, 100)
    y = np.linspace(-10, 10, 100)
    X, Y = np.meshgrid(x, y)
    Z = (-D - A * X - B * Y) / C
    ax.plot_surface(X, Y, Z, alpha=0.5, rstride=100, cstride=100, color='b')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    plt.title('Visualización de un Plano')
    plt.show()
    
# Ejemplo de uso
graficar_plano(1, -2, 1, 2)
```

```python
# Ejercicio 10
import numpy as np
import matplotlib.pyplot as plt

class Circle:
    def __init__(self, center, radius):
        self.center = center
        self.radius = radius
    def area(self):
        """Calcula el área del círculo."""
        return np.pi * self.radius ** 2
    def circumference(self):
        """Calcula la circunferencia del círculo."""
        return 2 * np.pi * self.radius
    def plot(self):
        """Dibuja el círculo usando matplotlib."""
        fig, ax = plt.subplots()
        theta = np.linspace(0, 2*np.pi, 100)
        x = self.center[0] + self.radius * np.cos(theta)
        y = self.center[1] + self.radius * np.sin(theta)
        ax.plot(x, y, 'b')  # Dibuja el círculo
        ax.plot(self.center[0], self.center[1], 'ro')  # Marca el centro
        ax.set_aspect('equal', adjustable='datalim')
        plt.grid(True)
        plt.show()
        
def main():
    center = (0, 0)
    radius = 1
    my_circle = Circle(center, radius)
    print("Área del círculo: {:.2f}".format(my_circle.area()))
    print("Circunferencia del círculo: {:.2f}".format(my_circle.circumference()))
    my_circle.plot()
    
if __name__ == "__main__":
    main()
```

```python
# Ejercicio 11
import numpy as np
import matplotlib.pyplot as plt

class Ellipse:
    def __init__(self, a, b):
        """
        Inicializa una nueva Elipse.
        :param a: Semieje mayor
        :param b: Semieje menor
        """
        self.a = a
        self.b = b
    def area(self):
        """
        Calcula el área de la elipse.
        :return: Área de la elipse.
        """
        return np.pi * self.a * self.b
    def circumference(self):
        """
        Aproxima la circunferencia de la elipse usando la fórmula de Ramanujan.
        :return: Circunferencia aproximada de la elipse.
        """
        h = ((self.a - self.b)**2) / ((self.a + self.b)**2)
        perimeter = np.pi * (self.a + self.b) * (1 + (3*h) / (10 + np.sqrt(4 - 3*h)))
        return perimeter
    def plot(self):
        """
        Grafica la elipse utilizando matplotlib.
        """
        theta = np.linspace(0, 2*np.pi, 100)
        x = self.a * np.cos(theta)
        y = self.b * np.sin(theta)
        plt.figure(figsize=(8, 6))
        plt.plot(x, y, label=f'Elipse con a={self.a}, b={self.b}')
        plt.scatter([0], [0], color='red')  # Centro de la elipse
        plt.title('Gráfico de la Elipse')
        plt.xlabel('Eje X')
        plt.ylabel('Eje Y')
        plt.axis('equal')
        plt.grid(True)
        plt.legend()
        plt.show()
        
# Ejemplo de uso
def main():
    a = 5  # longitud del semieje mayor
    b = 3  # longitud del semieje menor
    my_ellipse = Ellipse(a, b)
    print(f"Área de la elipse: {my_ellipse.area():.2f}")
    print(f"Perímetro de la elipse (aprox.): {my_ellipse.circumference():.2f}")
    my_ellipse.plot()
    
if __name__ == "__main__":
    main()
```

```python
# Ejercicio 12
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

class Cuadrado:
    def __init__(self, longitud_lado, centro=(0, 0)):
        """
        Inicializa un cuadrado con una longitud de lado dada y una posición central.
        :param longitud_lado: Longitud de los lados del cuadrado.
        :param centro: Una tupla (x, y) que representa el centro del cuadrado.
        """
        self.longitud_lado = longitud_lado
        self.centro = centro
    def area(self):
        """
        Calcula el área del cuadrado.
        :return: El área del cuadrado.
        """
        return self.longitud_lado ** 2
    def perimetro(self):
        """
        Calcula el perímetro del cuadrado.
        :return: El perímetro del cuadrado.
        """
        return 4 * self.longitud_lado
    def graficar(self):
        """
        Grafica el cuadrado usando matplotlib.
        """
        fig, ax = plt.subplots()
        esquina_inferior_izquierda = (self.centro[0] - self.longitud_lado / 2,
                                      self.centro[1] - self.longitud_lado / 2)
        
        # Crea un parche rectangular
        cuadrado = patches.Rectangle(esquina_inferior_izquierda, self.longitud_lado, self.longitud_lado, edgecolor='blue', facecolor='lightblue')
        ax.add_patch(cuadrado)
        
        # Establece límites para los ejes x e y
        ax.set_xlim(self.centro[0] - self.longitud_lado, self.centro[0] + self.longitud_lado)
        ax.set_ylim(self.centro[1] - self.longitud_lado, self.centro[1] + self.longitud_lado)
        
        # Establece que el aspecto de la gráfica sea igual
        ax.set_aspect('equal')
       
        # Añade cuadrícula, título y muestra
        ax.grid(True)
        plt.title('Gráfico del Cuadrado')
        plt.xlabel('Coordenada X')
        plt.ylabel('Coordenada Y')
        plt.show()

# Uso de ejemplo
def main():
    cuadrado = Cuadrado(2, centro=(0, 0))
    print("Área del cuadrado:", cuadrado.area())
    print("Perímetro del cuadrado:", cuadrado.perimetro())
    cuadrado.graficar()

if __name__ == "__main__":
    main()
```

```python
# Ejercicio 13
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

class Cuadrilatero:
    def __init__(self, vertices):
        """
        Inicializa un cuadrilátero con una lista de vértices.
        :param vertices: Lista de tuplas, donde cada tupla contiene las coordenadas (x, y) de un vértice.
        """
        self.vertices = np.array(vertices)
        self.lados = self.calcular_longitudes_lados()
    def calcular_longitudes_lados(self):
        """
        Calcula las longitudes de los lados del cuadrilátero basándose en los vértices.
        :return: Lista de longitudes de los lados.
        """
        lados = []
        num_vertices = len(self.vertices)
        for i in range(num_vertices):
            punto_inicio = self.vertices[i]
            punto_final = self.vertices[(i + 1) % num_vertices]
            longitud = np.linalg.norm(punto_inicio - punto_final)
            lados.append(longitud)
        return lados
    def area(self):
        """
        Calcula el área del cuadrilátero usando la fórmula de la coordenada del área.
        :return: Área del cuadrilátero.
        """
        x = self.vertices[:, 0]
        y = self.vertices[:, 1]
        return 0.5 * np.abs(np.dot(x, np.roll(y, 1)) - np.dot(y, np.roll(x, 1)))
    def perimetro(self):
        """
        Calcula el perímetro del cuadrilátero sumando las longitudes de los lados.
        :return: Perímetro del cuadrilátero.
        """
        return sum(self.lados)
    def graficar(self):
        """
        Grafica el cuadrilátero utilizando matplotlib.
        """
        fig, ax = plt.subplots()
        patch = patches.Polygon(self.vertices, closed=True, fill=None, edgecolor='r')
        ax.add_patch(patch)
        ax.set_xlim(min(self.vertices[:, 0]) - 1, max(self.vertices[:, 0]) + 1)
        ax.set_ylim(min(self.vertices[:, 1]) - 1, max(self.vertices[:, 1]) + 1)
        ax.set_aspect('equal')
        plt.grid(True)
        plt.show()

# Ejemplo de uso
def main():
    vertices = [(1, 2), (4, 3), (5, 7), (2, 6)]
    cuadrilatero = Cuadrilatero(vertices)
    print("Área del cuadrilátero:", cuadrilatero.area())
    print("Perímetro del cuadrilátero:", cuadrilatero.perimetro())
    cuadrilatero.graficar()

if __name__ == "__main__":
    main()
```

```python
# Ejercicio 18
import numpy as np
import matplotlib.pyplot as plt

def producto_punto(vector1, vector2):
    return np.dot(vector1, vector2)
def calcular_angulo(vector1, vector2):
    producto = producto_punto(vector1, vector2)
    magnitudes = np.linalg.norm(vector1) * np.linalg.norm(vector2)
    return np.arccos(producto / magnitudes)
def proyeccion_vector(vector1, vector2):
    producto = producto_punto(vector1, vector2)
    magnitud2_cuadrado = np.dot(vector2, vector2)
    return (producto / magnitud2_cuadrado) * vector2
def graficar_vectores(vector1, vector2, proyeccion):
    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(111, projection='3d')

    # Vectores y proyección
    ax.quiver(0, 0, 0, *vector1, color='r', label='Vector 1', arrow_length_ratio=0.1)
    ax.quiver(0, 0, 0, *vector2, color='b', label='Vector 2', arrow_length_ratio=0.1)
    ax.quiver(0, 0, 0, *proyeccion, color='g', linestyle='dashed', label='Proyección de Vector 1 sobre Vector 2', arrow_length_ratio=0.1)

    # Configuración de los límites del gráfico
    all_vectors = np.array([vector1, vector2, proyeccion])
    max_length = np.max(np.abs(all_vectors)) * 1.1  # Un poco más grande para asegurar visibilidad
    ax.set_xlim([-max_length, max_length])
    ax.set_ylim([-max_length, max_length])
    ax.set_zlim([-max_length, max_length])

    # Mejoras estéticas
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    plt.title('Visualización de Producto Punto y Proyección Vectorial')
    plt.legend()
    plt.grid(True)
    plt.show()

# Ejemplo de uso
def main():
    vector_a = np.array([2, 7, 5])
    vector_b = np.array([1, 1, 1])
    proyeccion_ab = proyeccion_vector(vector_a, vector_b)
    print("Producto punto:", producto_punto(vector_a, vector_b))
    angulo = calcular_angulo(vector_a, vector_b)
    print("Ángulo entre los vectores (radianes):", angulo)
    print("Proyección del vector A sobre B:", proyeccion_ab)
    graficar_vectores(vector_a, vector_b, proyeccion_ab)

if __name__ == "__main__":
    main()
```

```python
# Ejercicio 19
import numpy as np
import matplotlib.pyplot as plt

def son_paralelas_2d(linea1, linea2):
    """
    Determina si dos líneas en 2D son paralelas y las grafica.
    Las líneas están representadas por sus coeficientes (a, b, c) en la forma ax + by + c = 0.
    :param linea1: Coeficientes (a, b, c) de la primera línea.
    :param linea2: Coeficientes (a, b, c) de la segunda línea.
    :return: True si son paralelas, False de lo contrario.
    """
    a1, b1, c1 = linea1
    a2, b2, c2 = linea2
    paralelas = np.isclose(a1 * b2, a2 * b1)

    # Graficar líneas para visualizar
    graficar_lineas(linea1, linea2)
    return paralelas

def graficar_lineas(linea1, linea2):
    """
    Grafica dos líneas en un plano 2D para visualizar su relación.
    :param linea1: Coeficientes (a, b, c) de la primera línea.
    :param linea2: Coeficientes (a, b, c) de la segunda línea.
    """
    x = np.linspace(-10, 10, 400)
    y1 = (-linea1[0] * x - linea1[2]) / linea1[1] if linea1[1] != 0 else (linea1[2] / linea1[0])
    y2 = (-linea2[0] * x - linea2[2]) / linea2[1] if linea2[1] != 0 else (linea2[2] / linea2[0])
    plt.figure(figsize=(8, 6))
    plt.plot(x, y1, label=f'Línea 1: {linea1[0]}x + {linea1[1]}y + {linea1[2]} = 0')
    plt.plot(x, y2, label=f'Línea 2: {linea2[0]}x + {linea2[1]}y + {linea2[2]} = 0', linestyle='--')
    plt.xlim([-10, 10])
    plt.ylim([-10, 10])
    plt.axhline(0, color='grey', lw=0.5)
    plt.axvline(0, color='grey', lw=0.5)
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.title('Visualización de Líneas Paralelas con Diferentes Constantes')
    plt.legend()
    plt.grid(True)
    plt.show()

# Ejemplo de uso
def main():
    linea1 = (2, 3, -1)
    linea2 = (2, 3, -5)  # Misma pendiente, diferente constante
    paralelas = son_paralelas_2d(linea1, linea2)
    print("Las líneas son paralelas?:", paralelas)

if __name__ == "__main__":
    main()
```

```python
# Ejercicio 20
import numpy as np
import matplotlib.pyplot as plt

def verificar_ortogonalidad(v1, v2):
    """Verifica si dos vectores son ortogonales."""
    producto_punto = np.dot(v1, v2)
    return np.isclose(producto_punto, 0)
def es_matriz_ortogonal(matriz):
    """Verifica si una matriz es ortogonal."""
    Identidad = np.eye(matriz.shape[0])  # Matriz identidad del tamaño adecuado
    producto = np.dot(matriz.T, matriz)
    return np.allclose(producto, Identidad)
def graficar_vectores(vectores, titulo='Vectores en el espacio'):
    """Grafica una lista de vectores en 3D para visualizar la ortogonalidad."""
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    origen = np.zeros(3)
    for vector in vectores:
        ax.quiver(*origen, *vector, arrow_length_ratio=0.1)
    max_val = np.max(np.abs(vectores))
    ax.set_xlim([-max_val, max_val])
    ax.set_ylim([-max_val, max_val])
    ax.set_zlim([-max_val, max_val])
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    plt.title(titulo)
    plt.grid(True)
    plt.show()

# Ejemplo de uso
def main():
    v1 = np.array([1, 0, 0])
    v2 = np.array([0, 1, 0])
    v3 = np.array([0, 0, 1])
    print("v1 y v2 son ortogonales:", verificar_ortogonalidad(v1, v2))
    print("v2 y v3 son ortogonales:", verificar_ortogonalidad(v2, v3))
    print("v1 y v3 son ortogonales:", verificar_ortogonalidad(v1, v3))
    
    # Matriz ortogonal (ejemplo simple es la matriz identidad)
    matriz_identidad = np.eye(3)
    print("La matriz identidad es ortogonal:", es_matriz_ortogonal(matriz_identidad))
    
    # Graficar vectores
    graficar_vectores([v1, v2, v3], 'Vectores ortogonales básicos')

if __name__ == "__main__":
    main()
```

```python
# Ejercicio 21
import numpy as np
import matplotlib.pyplot as plt

def proyeccion_ortogonal(v, u):
    """
    Calcula la proyección ortogonal del vector v sobre el vector u.
    :param v: Vector a proyectar.
    :param u: Vector sobre el cual se proyecta.
    :return: Proyección ortogonal de v sobre u.
    """
    u_dot_v = np.dot(u, v)
    u_dot_u = np.dot(u, u)
    return (u_dot_v / u_dot_u) * u
def proyeccion_paralela(v, d):
    """
    Calcula la proyección paralela del vector v a lo largo de la dirección d.
    :param v: Vector a proyectar.
    :param d: Dirección de la proyección (vector unitario).
    :return: Proyección paralela de v a lo largo de d.
    """
    return np.dot(v, d) * d
def graficar_vectores(vectores, colores, titulo):
    """
    Grafica vectores en 2D para visualizar relaciones entre ellos.
    :param vectores: Lista de vectores a graficar.
    :param colores: Colores para cada vector.
    :param titulo: Título del gráfico.
    """
    plt.figure()
    for vector, color in zip(vectores, colores):
        plt.quiver(0, 0, vector[0], vector[1], angles='xy', scale_units='xy', scale=1, color=color)
    plt.xlim(-2, 10)
    plt.ylim(-2, 10)
    plt.axhline(0, color='black',linewidth=0.5)
    plt.axvline(0, color='black',linewidth=0.5)
    plt.grid(True)
    plt.title(titulo)
    plt.show()

# Proyección Ortogonal
v = np.array([8, 4])
u = np.array([2, 2])
proj_ortogonal = proyeccion_ortogonal(v, u)
print("Proyección Ortogonal:", proj_ortogonal)
graficar_vectores([v, u, proj_ortogonal], ['blue', 'green', 'red'], 'Proyección Ortogonal')

# Proyección Paralela
d = np.array([1, 0])  # Dirección unitaria para la proyección paralela
proj_paralela = proyeccion_paralela(v, d)
print("Proyección Paralela:", proj_paralela)
graficar_vectores([v, d, proj_paralela], ['blue', 'green', 'red'], 'Proyección Paralela')
```
