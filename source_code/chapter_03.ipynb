```python
# Ejercicio 2
import sympy as sp

# Ejemplo: g(x)=x^2
# Despejamos x para obtener: x = Sqrt(y)
# Dado que la raíz cuadrada de un número real siempre es real y no hay otras restricciones, el codominio es C_f :para todo y >= 0.
print("\nDespeje de x para el Ejemplo: g(x)=x^2")
y = sp.Symbol('y')
x_g = sp.sqrt(y)
codomain_g = sp.Interval(0, sp.oo)
print(f"C_f: {codomain_g}")

# Ejemplo: H(x)=1/x 
# Despejamos x para obtener: x = 1/y .
# Aquí debemos evitar la división por cero, por lo que y diferente de 0. Además, como la función representa la inversa de x, todos los valores reales excepto cero son válidos. 
# Por lo tanto, el co-dominio es: C_f : R-{0}.
print("\nDespeje de x para el Ejemplo: H(x)=1/x")
x_h = 1 / y
codomain_h = sp.Union(sp.Interval.open(-sp.oo, 0), sp.Interval.open(0, sp.oo))
print(f"C_f: {codomain_h}")

# Ejemplo: M(x)=Log(x+2)
# Despejamos x para obtener: x = e^y - 2.
# En este caso, debemos asegurarnos de que el argumento del logaritmo sea mayor que cero, por lo que e^y - 2 > 0. Resolviendo esta desigualdad, encontramos que y > ln(2).
# Por lo tanto, el codominio es C_f : (ln(2),infinito).
print("\nDespeje de x para el Ejemplo: M(x)=Log(x+2)")
x_m = sp.exp(y) - 2
codomain_m = sp.Interval.open(sp.log(2), sp.oo)
print(f"C_f: {codomain_m}")

# Ejemplo: N(x)=Sqrt(x-1) 
# Despejamos x para obtener: x = y^2 + 1.
# Aquí, para que la raíz cuadrada sea real, necesitamos que y^2 + 1 >= 0, lo cual siempre se cumple. Por lo tanto, el codominio es C_f : R.
print("\nDespeje de x para el Ejemplo: N(x)=Sqrt(x-1)")
x_n = y**2 + 1
codomain_n = sp.Interval(-sp.oo, sp.oo)
print(f"C_f: {codomain_n}")

# Ejemplo: P(x)=1/Sqrt((4-x) )
# Despejamos x para obtener: x = 4 - 1/y^2 .
# Similar al ejemplo anterior, el denominador de la fracción no puede ser cero, por lo que y^2 != 0, lo cual siempre se cumple. 
# Además, el argumento del denominador de la fracción debe ser mayor que cero para que la raíz cuadrada sea real, es decir, 4 - 1/y^2 > 0. 
# Resolviendo esta desigualdad, encontramos que y^2 < 1/4, lo cual se traduce en -1/2 < y < 1/2. 
print("\nDespeje de x para el Ejemplo: P(x)=1/Sqrt((4-x))")
x_p = 4 - 1 / y**2
codomain_p = sp.Interval.open(-1/2, 1/2)
print(f"C_f: {codomain_p}")
```

```python
# Ejercicio 6
import matplotlib.pyplot as plt
import networkx as nx

# Definir la posición de cada nodo para ser utilizada en el gráfico
pos = {1: (0, 1), 2: (0, 0.5), 3: (0, 0), 4: (1, 1), 5: (1, 0.5), 6: (1, 0)}

# Definir la función para dibujar el gráfico de función específico solicitado por el usuario
def dibujar_grafico_funcion_especifico(edges, pos, ax):
    # Inicializar un grafo dirigido
    G = nx.DiGraph()
    
    # Agregar la lista de aristas al grafo
    G.add_edges_from(edges)
    
    # Dibujar las aristas con puntas de flecha
    nx.draw_networkx_edges(G, pos, ax=ax, arrowstyle="->", arrowsize=20)
    
    # Etiquetar los nodos con números
    nx.draw_networkx_labels(G, pos, font_size=16, ax=ax)
    
    # Dibujar elipses para representar los conjuntos
    ellipse1 = plt.matplotlib.patches.Ellipse((0, 0.5), 0.5, 1.5, edgecolor='black', facecolor='none')
    ellipse2 = plt.matplotlib.patches.Ellipse((1, 0.5), 0.5, 1.5, edgecolor='black', facecolor='none')
    
    # Dibujar nodos circulares
    for nodo in pos:
        circulo = plt.Circle(pos[nodo], radius=0.15, edgecolor='black', facecolor='white')
        ax.add_patch(circulo)
    
    # Agregar las elipses al gráfico
    ax.add_patch(ellipse1)
    ax.add_patch(ellipse2)
    
    # Establecer los límites y quitar los ejes para una apariencia más limpia
    ax.set_xlim(-0.5, 1.5)
    ax.set_ylim(-0.5, 1.5)
    ax.xaxis.set_visible(False)
    ax.yaxis.set_visible(False)
    
# Definir las aristas para mostrar la dirección correcta de 1 a 2 y de 3 a 4
edges_corrected = [(1, 4), (2, 5), (3, 6)]

# Redibujar el gráfico de función específico con las aristas corregidas
fig, ax = plt.subplots(figsize=(6, 6))
dibujar_grafico_funcion_especifico(edges_corrected, pos, ax)
plt.show()
```

```python
# Ejercicio 13
import numpy as np
import matplotlib.pyplot as plt

# Redefinimos las funciones, incluyendo la corrección para la función logarítmica
funciones = {
    'f1': lambda x: x**2,                      # Función cuadrática
    'f2': lambda x: np.cos(x),                 # Función coseno
    'f3': lambda x: np.exp(x),                 # Función exponencial
    'f4': lambda x: np.where(x > 0, np.log(x), 0), # Función logarítmica con manejo de valores no positivos
    'f5': lambda x: 1 / (x**2 + 1),            # Función racional
    'f6': lambda x: np.abs(x),                 # Función valor absoluto
    'f7': lambda x: x**3,                      # Función cúbica
    'f8': lambda x: 2**x                       # Función exponencial base 2
}

# Creamos una figura grande para acomodar todas las gráficas
fig, axs = plt.subplots(4, 2, figsize=(15, 20))

# Generamos y graficamos los puntos para cada función
for ax, (nombre, funcion) in zip(axs.flatten(), funciones.items()):
    
    # Generamos 8 puntos para la función actual
    x_values = np.linspace(-2, 2, 8)
    y_values = funcion(x_values)
    
    # Graficamos los puntos
    ax.plot(x_values, y_values, 'o', label=f'Puntos de {nombre}')
    
    # Graficamos la función con una línea continua para mayor claridad
    x_func = np.linspace(min(x_values), max(x_values), 100)
    y_func = funcion(x_func)
    ax.plot(x_func, y_func, label=f'{nombre}(x)', color='gray')
    
    # Estilizamos el gráfico
    ax.axhline(y=0, color='k')  # Eje X
    ax.axvline(x=0, color='k')  # Eje Y
    ax.grid(True, which='both')
    ax.legend()
    
# Ajustamos el layout
plt.tight_layout()
plt.show()
```

```python
# Ejercicio 14
import numpy as np

# Función para verificar si una función es inyectiva
def es_inyectiva(func, rango):
    valores_vistos = set()
    for x in rango:
        y = func(x)
        if y in valores_vistos:
            return False  # No es inyectiva si encontramos el mismo valor de y para diferentes x
        valores_vistos.add(y)
    return True

# Definición de la función lineal f(x) = 2x
def f(x):
    return 2 * x

# Definición de la función cuadrática g(x) = x^2
def g(x):
    return x**2

# Rango de prueba
rango_prueba = np.linspace(-10, 10, 1000)

# Verificar la inyectividad de f(x) = 2x
resultado_f = es_inyectiva(f, rango_prueba)

# Verificar la inyectividad de g(x) = x^2
resultado_g = es_inyectiva(g, rango_prueba)
print("La función f(x) = 2x es inyectiva:", resultado_f)
print("La función g(x) = x^2 es inyectiva:", resultado_g)
```

```python
# Ejercicio 20
import numpy as np
import matplotlib.pyplot as plt

# Definición de la función original y sus transformaciones
def f_original(x):
    return x**2  # Ejemplo con una función cuadrática
def desplazamiento_derecha(x, c):
    return f_original(x - c)  # Desplazamiento c unidades a la derecha
def desplazamiento_arriba(x, c):
    return f_original(x) + c  # Desplazamiento c unidades arriba
def reflejo_eje_x(x):
    return -f_original(x)  # Reflejo sobre el eje X
def reflejo_eje_y(x):
    return f_original(-x)  # Reflejo sobre el eje Y

# Rango de prueba
x_values = np.linspace(-3, 3, 400)
c = 1  # Valor de desplazamiento y reflejo

# Gráficas de las transformaciones
plt.figure(figsize=(12, 8))

# Desplazamiento hacia la derecha
plt.subplot(2, 2, 1)
plt.plot(x_values, f_original(x_values), label="Original: f(x) = x^2")
plt.plot(x_values, desplazamiento_derecha(x_values, c), label="Desplazamiento Derecha", linestyle='dashed')
plt.title("Desplazamiento Derecha")
plt.legend()

# Desplazamiento hacia arriba
plt.subplot(2, 2, 2)
plt.plot(x_values, f_original(x_values), label="Original: f(x) = x^2")
plt.plot(x_values, desplazamiento_arriba(x_values, c), label="Desplazamiento Arriba", linestyle='dashed')
plt.title("Desplazamiento Arriba")
plt.legend()

# Reflejo sobre el eje X
plt.subplot(2, 2, 3)
plt.plot(x_values, f_original(x_values), label="Original: f(x) = x^2")
plt.plot(x_values, reflejo_eje_x(x_values), label="Reflejo Eje X", linestyle='dashed')
plt.title("Reflejo Eje X")
plt.legend()

# Reflejo sobre el eje Y
plt.subplot(2, 2, 4)
plt.plot(x_values, f_original(x_values), label="Original: f(x) = x^2")
plt.plot(x_values, reflejo_eje_y(x_values), label="Reflejo Eje Y", linestyle='dashed')
plt.title("Reflejo Eje Y")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Ejercicio 25
# Definición de las funciones cuadráticas y cálculo de sus raíces
def f_cuadratica1(x):
    return x**2 - 4*x + 3
def f_cuadratica2(x):
    return -x**2 + 6*x - 8
def f_cuadratica3(x):
    return x**2 + x + 1
def f_cuadratica4(x):
    return 2*x**2 - 8*x + 6
def calcular_raices(a, b, c):
    discriminante = b**2 - 4*a*c
    if discriminante >= 0:
        raiz1 = (-b + np.sqrt(discriminante)) / (2*a)
        raiz2 = (-b - np.sqrt(discriminante)) / (2*a)
        return [raiz1, raiz2]
    else:
        return []  # Sin raíces reales
    
# Rango de valores para x
x_values = np.linspace(-10, 10, 400)

# Raíces de las funciones cuadráticas
raices1 = calcular_raices(1, -4, 3)
raices2 = calcular_raices(-1, 6, -8)
raices3 = calcular_raices(1, 1, 1)  # Esta función no tiene raíces reales
raices4 = calcular_raices(2, -8, 6)

# Creación de gráficos
plt.figure(figsize=(12, 12))

# Gráfico de la primera función cuadrática
plt.subplot(2, 2, 1)
plt.axhline(0, color='red', linewidth=0.5)  # Eje Y
plt.axvline(0, color='red', linewidth=0.5)  # Eje X
plt.plot(x_values, f_cuadratica1(x_values), label="f(x) = x^2 - 4x + 3")
plt.scatter(raices1, [f_cuadratica1(x) for x in raices1], color='red')  # Raíces
plt.title("Parábola Abierta hacia Arriba")
plt.legend()

# Gráfico de la segunda función cuadrática
plt.subplot(2, 2, 2)
plt.axhline(0, color='red', linewidth=0.5)  # Eje Y
plt.axvline(0, color='red', linewidth=0.5)  # Eje X
plt.plot(x_values, f_cuadratica2(x_values), label="f(x) = -x^2 + 6x - 8")
plt.scatter(raices2, [f_cuadratica2(x) for x in raices2], color='red')  # Raíces
plt.title("Parábola Abierta hacia Abajo")
plt.legend()

# Gráfico de la tercera función cuadrática
plt.subplot(2, 2, 3)
plt.axhline(0, color='red', linewidth=0.5)  # Eje Y
plt.axvline(0, color='red', linewidth=0.5)  # Eje X
plt.plot(x_values, f_cuadratica3(x_values), label="f(x) = x^2 + x + 1")

# No se grafican raíces ya que no existen en el plano real
plt.title("Sin Raíces Reales")
plt.legend()

# Gráfico de la cuarta función cuadrática
plt.subplot(2, 2, 4)
plt.axhline(0, color='red', linewidth=0.5)  # Eje Y
plt.axvline(0, color='red', linewidth=0.5)  # Eje X
plt.plot(x_values, f_cuadratica4(x_values), label="f(x) = 2x^2 - 8x + 6")
plt.scatter(raices4, [f_cuadratica4(x) for x in raices4], color='red')  # Raíces
plt.title("Parábola Más Estrecha")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Ejercicio 36
import numpy as np
import matplotlib.pyplot as plt

# Definición de las funciones trigonométricas inversas
def arcsin(x):
    return np.arcsin(x)
def arccos(x):
    return np.arccos(x)
def arctan(x):
    return np.arctan(x)

# Creando un rango de valores para x que estén dentro del dominio de las funciones inversas
x_arcsin_arccos = np.linspace(-1, 1, 1000)  # Rango válido para arcsin y arccos
x_arctan = np.linspace(-10, 10, 1000)       # Rango más amplio para arctan debido a su dominio más extenso

# Creación de gráficos
plt.figure(figsize=(12, 4))

# Graficando la función arcoseno
plt.subplot(1, 3, 1)
plt.plot(x_arcsin_arccos, arcsin(x_arcsin_arccos), label='arcsin(x)')
plt.title('Función Arcoseno')
plt.xlabel('x')
plt.ylabel('arcsin(x)')
plt.legend()

# Graficando la función arcocoseno
plt.subplot(1, 3, 2)
plt.plot(x_arcsin_arccos, arccos(x_arcsin_arccos), label='arccos(x)')
plt.title('Función Arcocoseno')
plt.xlabel('x')
plt.ylabel('arccos(x)')
plt.legend()

# Graficando la función arcotangente
plt.subplot(1, 3, 3)
plt.plot(x_arctan, arctan(x_arctan), label='arctan(x)')
plt.title('Función Arcotangente')
plt.xlabel('x')
plt.ylabel('arctan(x)')
plt.legend()
plt.tight_layout()
plt.show()
```
