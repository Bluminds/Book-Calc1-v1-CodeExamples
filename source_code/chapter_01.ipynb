```python
# Ejercicio 3
import sympy as sp

# Definir la variable simbólica
x = sp.symbols('x')

# Definir la ecuación lineal: 20x + 3 = 7
ecuacion = sp.Eq(20*x + 3, 7)

# Resolver la ecuación para x
solucion = sp.solve(ecuacion, x)

# Mostrar la solución
print(f"Solución de la ecuación: x = {solucion[0]}")
```

```python
# Ejercicio 5
import sympy as sp

# Definir la variable simbólica
x = sp.symbols('x')

# Definir la expresión: x^3 + 8
expresion = x**3 + 8

# Factorizar la expresión
factorizada = sp.factor(expresion)

# Simplificar la expresión
simplificada = sp.simplify(expresion)

# Mostrar resultados
print("Expresión Original:")
display(expresion)

print("Expresión Factorizada:")
display(factorizada)

print("Expresión Simplificada:")
display(simplificada)
```

```python
# Ejercicio 6
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Definir la variable simbólica x
x = sp.Symbol('x')

# Definir el rango de valores de x
x_vals = np.linspace(0, 2, 400)  # Rango de 0 a 2 con 400 puntos

# Definir las funciones
f_1 = sp.lambdify(x, x**2 + 3*x - 2)
f_2 = sp.lambdify(x, 1/(x + 1))
f_3 = sp.lambdify(x, sp.exp(-x))

# Calcular los valores de y para cada función
y_vals_1 = f_1(x_vals)
y_vals_2 = f_2(x_vals)
y_vals_3 = f_3(x_vals)

# Graficar las funciones en un mismo gráfico
plt.figure(figsize=(10, 8))

# Función 1: f(x) = x^2 + 3x - 2
plt.plot(x_vals, y_vals_1, label='$f(x) = x^2 + 3x - 2$', color='red')

# Función 2: g(x) = 1/(x + 1)
plt.plot(x_vals, y_vals_2, label='$g(x) = \\frac{1}{x+1}$', color='green')

# Función 3: h(x) = e^(-x)
plt.plot(x_vals, y_vals_3, label='$h(x) = e^{-x}$', color='blue')

# Configuración del gráfico
plt.title('Gráficas de $f(x)$, $g(x)$ y $h(x)$ en el intervalo [0, 2]')
plt.xlabel('x')
plt.ylabel('y')
plt.xlim(0, 2)  # Establecer límites en el eje x
plt.axhline(y=0, color='black', linewidth=0.5)  # Línea horizontal y=0
plt.axvline(x=0, color='black', linewidth=0.5)  # Línea vertical x=0
plt.grid(True)
plt.legend()
plt.show()
```
