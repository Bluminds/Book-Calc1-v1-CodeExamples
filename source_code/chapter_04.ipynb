### 4.1.

```python
# Ejercicio 1
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# Función de ejemplo f(x) y sus parámetros
def f(x):
    return np.where(x != 0, np.sin(x) / x, 1)  # Evita división por cero

# Valores para el ejemplo
a = 1  # punto donde x tiende
L = f(a)  # Límite de la función en a
epsilon = 0.1  # Valor para ε (epsilon)
delta = 0.2  # Valor para δ (delta)

# Creando valores para x y y
x = np.linspace(a - delta - 1, a + delta + 1, 1000)
y = f(x)

# Creando el gráfico
plt.figure(figsize=(8, 6))
plt.plot(x, y, label="f(x) = sin(x)/x")

# Resaltando el intervalo alrededor de L en el eje Y
plt.fill_betweenx([L - epsilon, L + epsilon], a - delta, a + delta, color='blue', alpha=0.3)

# Resaltando el intervalo alrededor de a en el eje X
plt.axvspan(a - delta, a + delta, color='yellow', alpha=0.3)

# Marcando y anotando L, L+ε, L-ε, a-δ y a+δ
plt.scatter([a - delta, a, a - delta], [L - epsilon, L, L + epsilon], color='red')
plt.text(a - delta, L - epsilon, f'L-ε = {L - epsilon:.2f}', verticalalignment='bottom', horizontalalignment='right')
plt.text(a, L, f'L = {L:.2f}', verticalalignment='bottom', horizontalalignment='center')
plt.text(a + delta, L + epsilon, f'L+ε = {L + epsilon:.2f}', verticalalignment='bottom', horizontalalignment='right')
plt.text(a - delta, 0, f'a-δ = {a - delta:.2f}', verticalalignment='bottom', horizontalalignment='right')
plt.text(a + delta, 0, f'a+δ = {a + delta:.2f}', verticalalignment='bottom', horizontalalignment='left')

# Configuración del gráfico
plt.title('Gráfico del Análisis de Límites en el eje X y Y')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Ejercicio 2
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Definir la variable y la función
x = sp.symbols('x')
f = (x**2 - 1) / (x - 1)

# Aplicar la regla de L'Hopital
limit_f = sp.limit(f, x, 1)
print("El límite de la función original:", limit_f)

# Convertir la función a una función lambda para graficar
f_lambda = sp.lambdify(x, f)

# Definir valores de x cercanos a 1
x_values = np.linspace(0.5, 1.5, 1000)
y_values = f_lambda(x_values)

# Graficar la función
plt.plot(x_values, y_values, label='f(x)')
plt.axvline(x=1, color='r', linestyle='--', label='x=1 (Indeterminación)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gráfico de la función f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 15
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Definir la variable
x = sp.symbols('x')

# Definir la función g(x)
g = sp.tan(x) - 1/sp.cos(x) #NumPy no soporta sp.sec(x)

# Calcular el límite de g(x) cuando x tiende a pi/2 desde el lado izquierdo
limit_g_left = sp.limit(g, x, sp.pi/2, '-')
print("El límite de g(x) cuando x tiende a pi/2 desde el lado izquierdo:", limit_g_left)

# Calcular el límite de g(x) cuando x tiende a pi/2 desde el lado derecho
limit_g_right = sp.limit(g, x, sp.pi/2, '+')
print("El límite de g(x) cuando x tiende a pi/2 desde el lado derecho:", limit_g_right)

# Convertir g a una función lambda para graficar
g_lambda = sp.lambdify(x, g, "numpy")

# Definir un rango de valores para x
x_values = np.linspace(-1, np.pi, 1000)
x_values = x_values[(x_values < sp.pi/2 - 0.1) | (x_values > sp.pi/2 + 0.1)]

# Calcular los valores de g para los valores de x
y_values_g = g_lambda(x_values)

# Graficar la función
plt.figure(figsize=(8, 6))
plt.plot(x_values, y_values_g, label='g(x) = tan(x) - sec(x)')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.title('Gráfico de la función g(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 36
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Definición de la variable simbólica t y variacion_t
t, variacion_t = sp.symbols('t variacion_t')

# Definición de la función de posición s(t)
s = t**2

# Calculando el límite para la velocidad instantánea en t=2
v_t = sp.limit((s.subs(t, t + variacion_t) - s.subs(t, t)) / variacion_t, variacion_t, 0)
v_2 = v_t.subs(t, 2)  # Velocidad instantánea en t = 2 segundos

# Crear una función lambda para la representación gráfica
s_lambda = sp.lambdify(t, s, modules=['numpy'])

# Valores de t para la gráfica
t_vals = np.linspace(0, 4, 400)
s_vals = s_lambda(t_vals)

# Creando el gráfico
plt.figure(figsize=(8, 6))
plt.plot(t_vals, s_vals, label='s(t) = t^2')
plt.scatter(2, s_lambda(2), color='red')  # Marcar el punto t = 2
plt.annotate(f'v(2) = {v_2} m/s', (2, s_lambda(2)), textcoords="offset points", xytext=(0,10), ha='center')

# Configuración del gráfico
plt.title('Gráfico de s(t) = t^2 con Velocidad Instantánea en t=2')
plt.xlabel('Tiempo (t)')
plt.ylabel('Posición (s)')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()
```

```python
# Ejercicio 38
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Definición de la variable simbólica x
x = sp.symbols('x')

# Definición de la función f(x)
f = (2*x**2 + 3*x + 1) / (x + 1)

# Calculando el límite cuando x tiende a infinito para la asíntota oblicua
asintota_oblicua = sp.limit(f, x, sp.oo)

# Verificando la asíntota vertical en x = -1
lim_izq = sp.limit(f, x, -1, dir='-')
lim_der = sp.limit(f, x, -1, dir='+')

# Crear una función lambda para la representación gráfica
f_lambda = sp.lambdify(x, f, modules=['numpy'])

# Valores de x para la gráfica, evitando el punto de discontinuidad
x_vals1 = np.linspace(-10, -1.1, 400)
x_vals2 = np.linspace(-0.9, 10, 400)
f_vals1 = f_lambda(x_vals1)
f_vals2 = f_lambda(x_vals2)

# Creando el gráfico
plt.figure(figsize=(8, 6))
plt.plot(x_vals1, f_vals1, label='f(x) = (2x^2 + 3x + 1) / (x + 1)')
plt.plot(x_vals2, f_vals2, label='f(x) = (2x^2 + 3x + 1) / (x + 1)')
plt.axvline(x=-1, color='red', linestyle='--', label='Asíntota Vertical x=-1')
plt.plot(x_vals2, 2*x_vals2 + 3, 'g--', label='Asíntota Oblicua y=2x+3')

# Configuración del gráfico
plt.title('Gráfico de f(x) con Asíntotas')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()

print("Límite cuando x tiende a infinito (asíntota oblicua): y =", asintota_oblicua)
print("Límite cuando x tiende a -1 desde la izquierda:", lim_izq)
print("Límite cuando x tiende a -1 desde la derecha:", lim_der)
```

```python
# Ejercicio 39
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Definición de la variable simbólica t y variacion_t
t, variacion_t = sp.symbols('t variacion_t')

# Definición de la función g(t)
g = -16*t**2 + 64*t + 5

# Calculando la derivada de g(t) para encontrar los puntos extremos
g_prima = sp.diff(g, t)

# Resolviendo g'(t) = 0 para encontrar los puntos extremos
puntos_extremos = sp.solve(g_prima, t)

# Crear una función lambda para la representación gráfica
g_lambda = sp.lambdify(t, g, modules=['numpy'])

# Valores de t para la gráfica
t_vals = np.linspace(0, 5, 400)
g_vals = g_lambda(t_vals)

# Creando el gráfico
plt.figure(figsize=(8, 6))
plt.plot(t_vals, g_vals, label='g(t) = -16t^2 + 64t + 5')

# Marcando los puntos extremos en el gráfico
for punto in puntos_extremos:
    plt.scatter(punto, g_lambda(punto), color='red')
    plt.annotate(f't={punto.evalf():.2f}', (punto, g_lambda(punto)), textcoords="offset points", xytext=(0,10), ha='center')

# Configuración del gráfico
plt.title('Gráfico de g(t) = -16t^2 + 64t + 5 con Puntos Extremos')
plt.xlabel('Tiempo (t)')
plt.ylabel('Altura (g)')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()
```
