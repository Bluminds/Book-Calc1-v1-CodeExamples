```python
# Ejercicio 1
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import warnings
warnings.filterwarnings("ignore")

# Definir la variable simbólica y la función
x, h = sp.symbols('x h')
f = 4*x**2 + 10*x + 1

# Derivada usando el límite
f_prime = sp.limit((f.subs(x, x + h) - f) / h, h, 0)

# Convertir a funciones lambda para evaluar numéricamente
f_lamb = sp.lambdify(x, f, 'numpy')
f_prime_lamb = sp.lambdify(x, f_prime, 'numpy')

# Definir valores para x y un pequeño h para x+h
x_values = np.linspace(-10, 10, 400)
h_value = 0.5  # pequeño incremento para h
f_values = f_lamb(x_values)
f_prime_values = f_prime_lamb(x_values)
f_h_values = f_lamb(x_values + h_value)

# Crear gráfico
plt.figure(figsize=(12, 8))
plt.plot(x_values, f_values, label=r'$f(x) = 4x^2 + 10x + 1$', color='blue')
plt.plot(x_values, f_h_values, label=r'$f(x+h)$', linestyle='--', color='green')
plt.scatter([2], [f_lamb(2)], color='red', zorder=5)
plt.scatter([2 + h_value], [f_lamb(2 + h_value)], color='orange', zorder=5)
plt.text(2, f_lamb(2), 'f(x) en x', verticalalignment='bottom', horizontalalignment='right')
plt.text(2 + h_value, f_lamb(2 + h_value), 'f(x+h) en x+h', verticalalignment='top', horizontalalignment='left')

# Anotar derivada
plt.plot(x_values, f_prime_values, label=r"Derivada $f'(x) = 8x + 10$", color='magenta', linestyle='dotted')

# Mejorar el gráfico
plt.title('Función y su Derivada por Definición de Límite')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Ejercicio 6
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

def f(x):
    return np.exp(x)
def fractional_derivative_RL(a, x, alpha, func, n):
    """Calcula la derivada fraccionaria de Riemann-Liouville de orden alpha."""
    dx = 0.001
    t_values = np.arange(a, x, dx)
    integral = np.sum(func(t_values) / (x - t_values)**(alpha - n + 1)) * dx
    derivative = (1 / gamma(n - alpha)) * integral
    return derivative
def fractional_derivative_Caputo(a, x, alpha, func, n):
    """Calcula la derivada fraccionaria de Caputo de orden alpha."""
    dx = 0.001
    t_values = np.arange(a, x, dx)
    integral = np.sum(np.diff(func(t_values), n=n) / (x - t_values[:-n])**(-alpha + n + 1)) * dx
    derivative = (1 / gamma(-alpha + n + 1)) * integral
    return derivative

# Parámetros para la demostración
a = 0  # Punto inicial de integración
x_values = np.linspace(0.1, 2, 400)  # Puntos de evaluación
alpha = 0.5  # Orden fraccionario
n = 1  # Entero más pequeño mayor que alpha para Riemann-Liouville

# Calcular las derivadas fraccionarias para algunos puntos usando Riemann-Liouville
fractional_derivatives_RL = [fractional_derivative_RL(a, x, alpha, f, n) for x in x_values]

# Calcular las derivadas fraccionarias para algunos puntos usando Caputo (necesitamos ajustar este cálculo si se desea demostrar correctamente)
fractional_derivatives_Caputo = [fractional_derivative_Caputo(a, x, alpha, f, n) for x in x_values]

# Graficar las derivadas fraccionarias
plt.figure(figsize=(12, 6))
plt.plot(x_values, fractional_derivatives_RL, label="Riemann-Liouville Fractional Derivative", color="blue")
plt.plot(x_values, fractional_derivatives_Caputo, label="Caputo Fractional Derivative", color="green")
plt.title("Derivadas Fraccionarias de $e^x$")
plt.xlabel("x")
plt.ylabel("Derivative Value")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 8
import sympy as sp

def check_derivability(func, x0):
    x = sp.symbols('x')

    # Verificar la definición de la función en x0
    if func.subs(x, x0) is sp.nan:
        return False, "La función no está definida en x0."

    # Verificar el límite cuando x tiende a x0
    limit_left = sp.limit(func, x, x0, dir='-')
    limit_right = sp.limit(func, x, x0, dir='+')
    if limit_left != limit_right:
        return False, "El límite de la función cuando x tiende a x0 no existe."

    # Verificar continuidad en x0
    if func.subs(x, x0) != limit_left:
        return False, "La función no es continua en x0."

    # Verificar existencia de la derivada en x0
    derivative = sp.diff(func, x)
    derivative_at_x0 = sp.limit(derivative, x, x0)
    if derivative_at_x0 is sp.nan or derivative_at_x0.has(sp.oo, -sp.oo):
        return False, "La derivada no existe o no está definida en x0."
    return True, f"La función es derivable en x0 con una derivada de {derivative_at_x0}."

# Ejemplo de uso con una función cuadrática y un punto de interés x0 = 2
x = sp.symbols('x')
f = (x**3)+1
x0 = 2
result = check_derivability(f, x0)
result
```

```python
# Ejercicio 9
import sympy as sp

def check_diferentiability(func, x0):
    x = sp.symbols('x')

    # Verificar la definición de la función en x0
    if func.subs(x, x0) is sp.nan:
        return False, "La función no está definida en x0."

    # Verificar el límite cuando x tiende a x0
    limit_at_x0 = sp.limit(func, x, x0)
    if limit_at_x0 is sp.nan or limit_at_x0.has(sp.oo, -sp.oo):
        return False, "El límite de la función en x0 no existe o no está definido."

    # Verificar continuidad en x0
    if func.subs(x, x0) != limit_at_x0:
        return False, "La función no es continua en x0."

    # Verificar la existencia de la derivada en x0
    derivative = sp.diff(func, x)
    derivative_at_x0 = derivative.subs(x, x0)
    if derivative_at_x0 is sp.nan or derivative_at_x0.has(sp.oo, -sp.oo):
        return False, "La derivada no existe o no está definida en x0."

    # La función es diferenciable si es continua y tiene una derivada definida en x0
    return True, f"La función es diferenciable en x0. La derivada en x0 es {derivative_at_x0}."

# Ejemplo de uso con la función seno y un punto de interés x0 = pi/2
x = sp.symbols('x')
f = sp.sin(x)
x0 = sp.pi/2

result = check_diferentiability(f, x0)
result
```

```python
# Ejercicio 10
import sympy as sp

def display_derivative_rules():
    x, a, m, n = sp.symbols('x a m n')
    u = sp.Function('u')(x)
    v = sp.Function('v')(x)
    w = sp.Function('w')(x)
    
    # Reglas de derivación
    derivatives = {
        "Derivada de una constante": sp.diff(a, x),
        "Derivada de suma (u+v)": sp.diff(u + v, x),
        "Derivada de constante por función (au)": sp.diff(a * u, x),
        "Derivada de un producto (uv)": sp.diff(u * v, x),
        "Derivada de un cociente (u/v)": sp.diff(u / v, x),
        "Derivada de producto de tres funciones (uvw)": sp.diff(u * v * w, x),
        "Derivada de función elevada a función (u^v)": sp.diff(u**v, x),
        "Regla de la cadena (u(v(x)))": sp.diff(u.subs(u.args[0], v), x),
        "Derivada de función inversa": 1 / sp.diff(u, x),
        "Derivada de (x^m)": sp.diff(x**m, x),
        "Derivada de (a^x)": sp.diff(a**x, x),
        "Derivada de (e^x)": sp.diff(sp.exp(x), x),
        "Derivada de (Ln(x))": sp.diff(sp.log(x), x),
        "Derivada de (Log_a(x))": sp.diff(sp.log(x, a), x),
        "Derivada de (x^x)": sp.diff(x**x, x),
        "Derivada de (sen(x))": sp.diff(sp.sin(x), x),
        "Derivada de (cos(x))": sp.diff(sp.cos(x), x),
        "Derivada de (tg(x))": sp.diff(sp.tan(x), x),
        "Derivada de (ctg(x))": sp.diff(1/sp.tan(x), x),
        "Derivada de (sec(x))": sp.diff(1/sp.cos(x), x),
        "Derivada de (csc(x))": sp.diff(1/sp.sin(x), x),
        "Derivada de (senh(x))": sp.diff(sp.sinh(x), x),
        "Derivada de (cosh(x))": sp.diff(sp.cosh(x), x),
        "Derivada de (tgh(x))": sp.diff(sp.tanh(x), x),
        "Derivada de (ctgh(x))": sp.diff(1/sp.tanh(x), x),
        "Derivada de (sech(x))": sp.diff(1/sp.cosh(x), x),
        "Derivada de (csch(x))": sp.diff(1/sp.sinh(x), x)
    }
    
    # Imprimir las derivadas calculadas
    for rule, derivative in derivatives.items():
        print(f"{rule}: {derivative}")
        
display_derivative_rules()
```

```python
# Ejercicio 15
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff, solve, Eq

# Definir la variable x para SymPy
x = symbols('x')

# Definir la función
f = x**3 - 2*x**2 + 1

# Calcular la primera y segunda derivadas
f_prime = diff(f, x)
f_double_prime = diff(f_prime, x)

# Resolver f'(x) = 0 para encontrar puntos críticos
critical_points = solve(Eq(f_prime, 0), x)

# Evaluación de la segunda derivada en los puntos críticos
second_derivatives = {cp: f_double_prime.subs(x, cp) for cp in critical_points}

# Definir los valores de x para la gráfica
x_values = np.linspace(-1, 2, 400)
f_values_numeric = [f.subs(x, val).evalf() for val in x_values]  # Evaluación numérica de la función

# Crear la gráfica de la función
plt.figure(figsize=(8, 5))
plt.plot(x_values, f_values_numeric, label='f(x) = $x^3 - 2x^2 + 1$', color='blue')

# Marcar los puntos críticos con colores según si son máximos o mínimos
for cp, sd in second_derivatives.items():
    cp_y = f.subs(x, cp).evalf()  # Valor de f en el punto crítico
    color = 'red' if sd < 0 else 'green'  # Rojo para máximos, verde para mínimos
    plt.scatter([cp], [cp_y], color=color, zorder=5)
    plt.text(cp, cp_y, f'P({cp}, {cp_y:.2f})', fontsize=12, verticalalignment='bottom', horizontalalignment='right')

# Añadir título y etiquetas a la gráfica
plt.title('Gráfica de la función y extremos locales')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()

# Resultados de los puntos críticos y sus valores de segunda derivada
print("Puntos críticos y valores de la segunda derivada:")
for cp, sd in second_derivatives.items():
    print(f'Punto: x = {cp}, f\'\'(x) = {sd}, {"mínimo" if sd > 0 else "máximo"} local')
```

```python
# Ejercicio 21
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff, solve, Eq, sqrt, N, re, im

# Definir la variable x
x = symbols('x')

# Punto P y la parábola y = x^2 - 2x
P = (1, 3)
y = x**2 - 2*x

# Función de la distancia al cuadrado para simplificar la minimización
d_squared = (x - P[0])**2 + (y - P[1])**2

# Derivar la función de distancia al cuadrado
d_squared_prime = diff(d_squared, x)

# Encontrar puntos críticos resolviendo la derivada igual a cero
critical_points = solve(d_squared_prime, x)

# Segunda derivada para el test de concavidad
d_squared_double_prime = diff(d_squared_prime, x)

# Evaluar la segunda derivada en los puntos críticos, filtrando valores no reales
second_derivative_evals = {}
for cp in critical_points:
    second_deriv_val = N(d_squared_double_prime.subs(x, cp))
    
    # Asegurar que sólo tomamos valores reales y no complejos
    if im(second_deriv_val) == 0:
        second_derivative_evals[cp] = re(second_deriv_val)

# Filtrar solo los puntos donde la segunda derivada es positiva, indicando mínimos locales
min_points = {cp: val for cp, val in second_derivative_evals.items() if val > 0}

# Evaluar la distancia en esos puntos para determinar la distancia mínima
min_distances = {}
for cp in min_points:
    dist_val = sqrt(d_squared.subs(x, cp))
    
    # Convertir a un valor real si es necesario
    if im(dist_val) == 0:
        min_distances[cp] = N(dist_val)

# Puntos para graficar la parábola
x_vals = np.linspace(-2, 3, 400)
y_vals = [y.subs(x, val) for val in x_vals]

# Preparar la gráfica
plt.figure(figsize=(6, 6))
plt.plot(x_vals, y_vals, label='Parábola y = x^2 - 2x', color='blue')

# Marcar el punto P y los puntos críticos mínimos en la parábola
plt.scatter([P[0]], [P[1]], color='red', label='Punto P(1,3)', zorder=5)
for cp, dist in min_distances.items():
    cp_f   = float(cp)                           # x numérico
    y_cp_f = float(y.subs(x, cp))                # y numérico en la parábola
    dist_f = float(dist)                         # distancia numérica

    plt.scatter([cp_f], [y_cp_f], color='green', zorder=5)
    plt.text(cp_f, y_cp_f, f'x={cp_f:.2f}, d={dist_f:.2f} m',
             fontsize=12, va='bottom', ha='right')
    
# Configuración de la gráfica
plt.title('Distancia Mínima desde P(1,3) a la Parábola y = x^2 - 2x')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()

# Resultados de los puntos críticos y sus evaluaciones
print("Puntos críticos:", critical_points)
print("Evaluaciones de la segunda derivada:", second_derivative_evals)
print("Distancias mínimas:", min_distances)
```

```python
# Ejercicio 23
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff, sin, cos, pi, N

# Definir la variable x
x = symbols('x')

# Definir la función f(x) = x^2 - 4x + 6 en el intervalo [0, 5]
f = x**2 - 4*x + 6

# Calcular la derivada
f_prime = diff(f, x)

# Valores de la función en los extremos del intervalo [0, 5]
f_a = f.subs(x, 0)  # f(0)
f_b = f.subs(x, 5)  # f(5)

# Calcular la pendiente promedio entre los puntos [0, 5]
pendiente_promedio = (f_b - f_a) / (5 - 0)

# Resolver f'(c) = pendiente promedio para encontrar c
c_value = solve(Eq(f_prime, pendiente_promedio), x)

# Convertir c_value a un número flotante
c = float(N(c_value[0]))

# Puntos para graficar la función y la línea secante
x_vals = np.linspace(0, 5, 400)
f_vals = [f.subs(x, val).evalf() for val in x_vals]

# Valor de f en el punto c
f_c = f.subs(x, c)

# Preparar la gráfica
plt.figure(figsize=(10, 6))
plt.plot(x_vals, f_vals, label='f(x) = $x^2 - 4x + 6$', color='blue')

# Marcar el punto c y la línea secante
plt.scatter([0, 5, c], [f_a, f_b, f_c], color='red')  # Marcar los puntos a, b, y c
plt.title('Teorema del Valor Medio para f(x)')

# Dibujar la línea secante desde (0, f(0)) hasta (5, f(5))
line_x = np.array([0, 5])
line_y = np.array([f_a, f_b])
plt.plot(line_x, line_y, 'g--', label='Línea secante')

# Añadir la línea tangente en el punto c
tangent_line_y = f_prime.subs(x, c) * (x_vals - c) + f_c
plt.plot(x_vals, tangent_line_y, 'r--', label='Línea tangente en c')

# Añadir título, etiquetas y leyenda
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()

# Imprimir los valores importantes
f"El punto c encontrado es {c}, donde la derivada f'(c) = {f_prime.subs(x, c)} es igual a la pendiente promedio {pendiente_promedio}"
```

```python
# Ejercicio 24
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, exp, factorial

# Definir la variable x
x = symbols('x')

# Definir la función f(x) = e^x
f = exp(x)

# Coordenada en torno a la cual se desarrolla el polinomio de Taylor
a = 0

# Derivadas de f(x) = e^x en a = 0
f_0 = f.subs(x, a)  # f(0) = e^0 = 1
f_prime_0 = f.diff(x).subs(x, a)  # f'(0) = e^0 = 1
f_double_prime_0 = f.diff(x, 2).subs(x, a)  # f''(0) = e^0 = 1
f_triple_prime_0 = f.diff(x, 3).subs(x, a)  # f'''(0) = e^0 = 1

# Polinomio de Taylor de tercer grado para f(x) = e^x alrededor de a = 0
P_3 = (f_0 +
       f_prime_0 * (x - a) +
       f_double_prime_0 / factorial(2) * (x - a)**2 +
       f_triple_prime_0 / factorial(3) * (x - a)**3)

# Valores de x para graficar
x_vals = np.linspace(-2, 2, 400)
f_vals = [f.subs(x, val).evalf() for val in x_vals]
P_3_vals = [P_3.subs(x, val).evalf() for val in x_vals]

# Preparar la gráfica
plt.figure(figsize=(10, 6))
plt.plot(x_vals, f_vals, label='f(x) = e^x', color='blue', linewidth=2)
plt.plot(x_vals, P_3_vals, label='Polinomio de Taylor P_3(x)', color='green', linestyle='--', linewidth=2)

# Añadir título, etiquetas y leyenda
plt.title('Aproximación de f(x) = e^x usando Polinomio de Taylor de grado 3')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 25
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, sin, factorial, pi

# Definir la variable x
x = symbols('x')

# Definir la función f(x) = sin(x)
f = sin(x)

# Coordenada en torno a la cual se desarrolla el polinomio de Taylor
a = 0  # Expansión alrededor de x = 0

# Calculando las derivadas de f(x) = sin(x) en a = 0 hasta la décima derivada
derivadas_10_final = [f.diff(x, n).subs(x, a) for n in range(11)]

# Construir el polinomio de Taylor de décimo grado para f(x) = sin(x) alrededor de a = 0
P_10_final = sum([derivadas_10_final[n] / factorial(n) * (x - a)**n for n in range(11)])

# Valores de x para graficar
x_vals_10_final = np.linspace(-4 * float(pi), 4 * float(pi), 800)
f_vals_10_final = [f.subs(x, val).evalf() for val in x_vals_10_final]
P_10_vals_final = [P_10_final.subs(x, val).evalf() for val in x_vals_10_final]

# Preparar la gráfica
plt.figure(figsize=(12, 7))
plt.plot(x_vals_10_final, f_vals_10_final, label='f(x) = sin(x)', color='blue', linewidth=2)
plt.plot(x_vals_10_final, P_10_vals_final, label='Polinomio de Taylor P_10(x)', color='green', linestyle='--', linewidth=2)

# Configuración de los límites del eje y
plt.ylim(-2, 2)

# Añadir título, etiquetas y leyenda
plt.title('Aproximación de f(x) = sin(x) usando Polinomio de Taylor de grado 10')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 27
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff

# Definir la variable x
x = symbols('x')

# Definir la función f(x) = x^2
f = x**2

# Derivar f para obtener la pendiente de la recta tangente
f_prime = diff(f, x)

# Punto de tangencia
a = 4
f_a = f.subs(x, a)  # f(4) = 16
f_prime_a = f_prime.subs(x, a)  # f'(4) = 8

# Ecuación de la recta tangente
# y = f(a) + f'(a) * (x - a)
tangent_line = f_a + f_prime_a * (x - a)

# Valores de x para graficar
x_vals = np.linspace(0, 8, 400)
f_vals = [f.subs(x, val).evalf() for val in x_vals]
tangent_vals = [tangent_line.subs(x, val).evalf() for val in x_vals]

# Preparar la gráfica
plt.figure(figsize=(10, 6))
plt.plot(x_vals, f_vals, label='f(x) = x^2', color='blue', linewidth=2)
plt.plot(x_vals, tangent_vals, label=f'Recta Tangente en x={a}', color='red', linestyle='--', linewidth=2)

# Marcar el punto de tangencia
plt.scatter([a], [f_a], color='green', zorder=5)  # Punto (4, 16)
plt.text(a, f_a, f'({a}, {f_a})', fontsize=12, verticalalignment='bottom', horizontalalignment='right')

# Añadir título, etiquetas y leyenda
plt.title('Recta Tangente a la Curva f(x) = x^2 en x=4')
plt.xlabel('x')
plt.ylabel('f(x) / Recta Tangente')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 28
import numpy as np
import matplotlib.pyplot as plt

# Definir la función f(x) = x^2
def f(x):
    return x**2

# Definir la ecuación de la recta normal y = -1/8x + 16.5
def normal_line(x):
    return -1/8 * x + 16.5

# Definir la ecuación de la recta tangente y = 8x - 16
def tangent_line(x):
    return 8 * x - 16

# Punto donde se encuentra la tangente
x0 = 4
y0 = f(x0)

# Rango de valores de x
x_values = np.linspace(-20, 20, 400)

# Calcular los valores de y para la función f(x)
y_values = f(x_values)

# Calcular los valores de y para la recta normal y la recta tangente
normal_values = normal_line(x_values)
tangent_values = tangent_line(x_values)

# Configurar el tamaño del gráfico
plt.figure(figsize=(5, 6))

# Graficar la función f(x)
plt.plot(x_values, y_values, label='f(x) = x^2')

# Graficar la recta normal
plt.plot(x_values, normal_values, label='Recta Normal')

# Graficar la recta tangente
plt.plot(x_values, tangent_values, label='Recta Tangente')

# Marcar el punto donde se encuentra la tangente
plt.scatter(x0, y0, color='red', label=f'Punto (4, {y0})')

# Configuración de la gráfica
plt.xlabel('x')
plt.ylabel('y')
plt.title('Curvas y = x^2, Recta Normal y Recta Tangente')
plt.legend()
plt.grid(True)

# Limitar los ejes al rango de -20 a 20 en ambos ejes
plt.axis([-10, 10, -5, 30])

# Mostrar la gráfica
plt.show()
```

```python
# Ejercicio 29
import numpy as np
import matplotlib.pyplot as plt

# Definir la función f(x) = x^2
def f(x):
    return x**2

# Definir el punto de interés
x0 = 2

# Calcular la derivada de la función en el punto de interés
def f_prime(x):
    return 2 * x

# Calcular la pendiente de la tangente en el punto de interés
m = f_prime(x0)

# Calcular el ángulo que la tangente forma con el eje horizontal
theta = np.arctan(m)

# Convertir el ángulo de radianes a grados
theta_degrees = np.degrees(theta)

# Imprimir el resultado
print(f"El ángulo que la tangente forma con el eje horizontal en x={x0} es aproximadamente {theta_degrees:.2f} grados.")

# Configurar el tamaño del gráfico
plt.figure(figsize=(8, 6))

# Graficar la función f(x)
x_values = np.linspace(-15, 15, 200)
y_values = f(x_values)
plt.plot(x_values, y_values, label='f(x) = x^2')

# Graficar la tangente en el punto de interés
tangent_values = f(x0) + m * (x_values - x0)
plt.plot(x_values, tangent_values, linestyle='--', label='Tangente en x=2')

# Marcar el punto de interés
plt.scatter(x0, f(x0), color='red', label=f'Punto de interés (x={x0})')

# Configuración de la gráfica
plt.xlabel('x')
plt.ylabel('y')
plt.title('Ángulo de la tangente con el eje horizontal')
plt.legend()
plt.grid(True)

# Ajustar los ejes a la misma escala
plt.axis('equal')

# Mostrar la gráfica
plt.show()
```

```python
# Ejercicio 30
import numpy as np
import matplotlib.pyplot as plt

# Definir la función f(x) = x^2 - 2
def f(x):
    return x**2 - 2

# Definir la derivada de la función f(x) = x^2 - 2
def f_prime(x):
    return 2 * x

# Definir el valor inicial
x0 = 1

# Definir el número máximo de iteraciones
max_iterations = 20

# Definir la tolerancia para la convergencia
tolerance = 1e-6

# Listas para almacenar las iteraciones y los valores de x
iterations = []
x_values = []

# Iterar para encontrar la raíz utilizando el Método de Newton-Raphson
x = x0
for i in range(max_iterations):
    x_values.append(x)
    iterations.append(i)
    if abs(f(x)) < tolerance:
        break
    x = x - f(x) / f_prime(x)

# Imprimir la raíz aproximada y el número de iteraciones
print(f"Raíz aproximada: {x}")
print(f"Número de iteraciones: {i+1}")

# Graficar la función f(x)
x_range = np.linspace(-2, 2, 400)
plt.plot(x_range, f(x_range), label='f(x) = x^2 - 2')

# Graficar los puntos de cada iteración con su número de iteración
for i in range(len(x_values)):
    plt.scatter(x_values[i], f(x_values[i]), color='red')
    plt.text(x_values[i], f(x_values[i]), str(i+1), fontsize=9, verticalalignment='bottom', horizontalalignment='right')

# Configuración de la gráfica
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Iteraciones del Método de Newton-Raphson')
plt.legend()
plt.grid(True)

# Mostrar la gráfica
plt.show()
```
